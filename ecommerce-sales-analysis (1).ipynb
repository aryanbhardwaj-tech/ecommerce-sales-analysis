# Converting data column

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
```

```python
data.info()
```

```python
data.head()
```

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of week'] = data['Order Date'].dt.dayofweek
```

# monthly sales analysis

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
fig = px.line(sales_by_month,
              x ='Order Month',
              y ='Sales',
              title='Monthly Sales Analysis')
fig.show()
```

```python
data.head()
```

# sales by category

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
fig = px.pie(sales_by_category,
             values='Sales',
             names='Category',
             hole=0.5,
             color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(textposition='inside',textinfo='percent+label')
fig.update_layout(title_text='Sales Analysis by Category',title_font=dict(size=24))

fig.show()
```

# sales and profit - customer segment

```python
data.head()
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales':'sum','Profit': 'sum'}).reset_index()

color_palette = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],
                     y=sales_profit_by_segment['Sales'],
                     name='Sales',
                     marker_color=color_palette[0]))
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],
                     y=sales_profit_by_segment['Profit'],
                     name='Profit',
                     marker_color=color_palette[1]))

fig.update_layout(title='Sales and Profit Analysis by Customer Segment',
                  xaxis_title='Customer Segment',yaxis_title='Amount')

fig.show()
                     
                     
```

# sales to profit ratio

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales':'sum','Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales']/sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','Sales_to_Profit_Ratio']])
```

```python
sales_profit_by_segment['Profit_Margin_%'] = (
    sales_profit_by_segment['Profit'] / sales_profit_by_segment['Sales'] * 100
)
print(sales_profit_by_segment[['Segment', 'Profit_Margin_%']].round(2))
```

**Insight:** Consumer has the highest sales but the lowest profit margin (11.55%). Home Office has the lowest sales but the best margin (14.03%). Corporate sits in between (13.03%).

## Conclusion
- Overall: $2.30M sales and $286K profit (12.5% margin) across 2014–2017.
- Sales peak in Nov ($352K) and Dec ($325K) and are lowest in Feb; profit peaks in December.
- Technology is the top category in both sales and profit.
- Furniture has high sales but only ~2.5% margin; Tables, Bookcases and Supplies sell but lose money.
- Consumer is the largest segment but has the lowest margin (11.55%), while Home Office is the most profitable per sale (14.03%).

**Recommendation:** review discounts and pricing on Tables and Bookcases, and plan stock and marketing around the Sep–Dec peak.
